```r
library(palmerpenguins)

mutantPenguins <- penguins %>%
  mutate(bill_length_mm = bill_length_mm/25.4)%>%
  select(bill_length_mm,species,island,bill_depth_mm)

head(mutantPenguins)

mutantPenguins <- penguins %>%
  mutate(bill_length_in = bill_length_mm/25.4)%>%
  select(bill_length_mm,species,island,bill_depth_mm,bill_length_in)

head(mutantPenguins)
```

```r
meanBills <- penguins %>%
  select(bill_length_mm,bill_depth_mm)%>%
  summarize(
    mean_length = mean(bill_length_mm,na.rm=T),
    mean_depth=mean(bill_depth_mm, na.rm=T))

meanBills
```

```r
challenge1 <- starwars1 %>%
mutate(height = height/254, mass= mass*2.205)
challenge1
```

```r
challenge2 <- starwars1 %>%
group_by(homeworld) %>%
mutate(homeworld_pop=n())
challenge2
```

```r
challenge3 <- starwars1 %>%
summarize(
    name = n(),
    median_height = median(height, na.rm=T),
    distinct_homeworlds = n_distinct(homeworld)
)
challenge3
```

```r
challenge4 <- starwars1 %>%
group_by(gender) %>%
mutate(gender_pop=n())%>%
summarize(
median_female_height = median(height(female), na.rm = T),
median_male_height = median(height(male), na.rm = T)
)
challenge4
```

```r
challenge4 <- starwars1 %>%
group_by(gender) %>%
summarize(
total=n(),
median=median(height, na.rm=T)
)

challenge4
```
